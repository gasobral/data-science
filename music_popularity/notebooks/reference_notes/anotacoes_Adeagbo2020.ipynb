# Anotações do artigo 8
**Título:** [Predicting Afrobeats Hit Songs Using Spotify Data](https://arxiv.org/pdf/2007.03137)

**Autor:** Adewale Adeagbo

**Ano:** 2020

## Objetivo
Prever qual música, do gênero Afrobetas, será popular no Spotify.

## Metodologia

### Modelos utilizados
- Logistic Regression
- Decision Tree
- Random Forest Classifier
- Gradient Boosting - XGBoost
- Neural Network

### Data sets
Foram obtidas 2036 músicas Afrobeat usando a API do Spotify de 2010 a
2020.

### Limpeza dos dados

### Análise Exploratória
Foi criado uma variável chamada *hit* para indicar se uma música foi
popular ou não. Se a populariade (campo obtido da API do Spotify) > 47,
então hit = 1; caso contrário, hit = 0.

### Treinamento dos modelos
Foram utilizadas as seguintes proporções para treinar os modelos:

- 80% para treino e 20% para teste
- 75% para treino e 25% para validação

## Principais resultados
*Danceability*, *loudness* e *time_signature* foram as features com as
três maiores forças positivas com a target. Enquanto que,
*duration_ms*, *energy* e *valence* foram as três com as menores
forças.

Dos modelos utilizados, a floresta aleatória com XGBoost obteve os
melhores resultados, considerando todas as métricas.
