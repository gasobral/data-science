## Tratamento dos dados
- remove revendas não recorrentes (sugere algo no sentido de uma série
  temporal auto-correlacionada)
- análise individual de cada catálogo
- assume um crescimento linear
- Utilizaram Variance Against Forecast (VAF) e Pro-Forma Annual Revenue (PFAR) nas suas analises de dados
