## Metodologia
### Modelos utilizados
O autor propôs usar as *Redes Neurais*, mas ele também fez um estudo
comparativo com *Floresta Aleatória* e *SVM*.

### Limpeza dos dados
As seguintes ações foram realizadas:

- retirou as colunas com ID, pois não trazem informações relevantes
  para a análise

- usou a coluna *year* e retirou a *release date*, porque boa parte dos
  registros não possuem a data completa e boa parte dos valores de year
  são iguais as de release date.

> Observação:
> No nosso caso, prever o valor de passe, não seria interessante esse
> tipo de remoção de datas.

- foram removidas linhas que continham o mesmo nome e artista, mas
  diferentes versões (ao vivo, remaster ou outras) foram mantidas

### Análise Exploratória
Foi descoberto que boa parte das músicas tem popularidade 0, enquanto
que uma pequena parcela das música possuem popularidade acima de
80. Para contornar o problema de classe desbalanceada, o autor fez o
seguinte:

- removeu músicas com popularidade = 0 para que elas ficassem o mesmo
  número de músicas com popularidade = 42 (segunda maior popularidade)

- usou um cut-off para classificar uma música em popular ou não, ou
  seja, se popularidade > 40 então a música é popular, caso contrário,
  não é popular.

## Principais resultados
Sobre a revisão da literatura, Otuokere (Otuokere et al, 2021) faz uma
análise exploratória dos dados de 160.000 músicas do Spotify (data set
usado nesse trabalho). Talvez seja interessante dar uma olhada nesse
trabalho para ver se há algum insight relvante.

Outro trabalho interessante, mencionado pelo autor, é a tese do Essa
(Essa et al, 2022). Essa também utiliza o data set to Spotify e faz uma
análise usando 7 modelos de classificação e 4 de regressão. Nesse
trabalho, a floresta aleatória apresenta o melhor resultado. Inclusive,
Rusconi diz que o trabalho dele se diferencia desse na etapa de limpeza
dos dados. Rusconi utiliza um cut-off (41 de popularidade) para
estipular quando uma música é popular, enquanto que Essa usa a
popularidade (valor de 0 a 100) do Spotify.

Sobre os modelos:
- SVM foi o melhor algoritmo para classificar uma música em popular
- Redes Neurais foram o melhor algoritmo para identificar uma música em
  não popular
- Floresta Aleatória teve o melhor balanço entre performance e acurácia
- De modo geral, os algoritmos têm F1 score de 80%

## Limitações
