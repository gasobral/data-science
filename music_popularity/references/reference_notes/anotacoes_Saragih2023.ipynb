# Anotações do artigo 4
**Título:** [Predicting song popularity based on Spotify's audio features: insights from the Indonesian streaming users](https://www.tandfonline.com/doi/full/10.1080/23270012.2023.2239824#d1e114)

**Autor:** Harriman Samuel Saragih

**Ano:** 2023

## Objetivo
Prever a popularidade de uma música usando propriedades acústicas do
Spotify Indonésia e modelos de Regressão e Classificação.

> Observação:
> Ao ler a revisão da literatura, podemos notar que a localização tem
> uma influência nas *features* usadas nos modelos. De modo geral, a
> floresta aleatória apresenta uma boa performance (nesse trabalho e em
> outros).

## Metodologia
### Modelos utilizados
Linear Regression, Neural Network, Support Vector Regression e Random
Forest.

### Data sets
Foi utilizado um data set to Spotify Indonésia. Além disso, foi
apresentado um dicionário dos dados.

### Pipeline dos dados
No tratamento dos dados foram realizados as seguintes etapas:

1. Pré-processamento (remoção de outliers, dados ausentes, remoção de
variáveis não numéricas e one-hot-encoding de variáveis categóricas com
*n* categorias)
2. Treino-Teste
3. Validação

No Pré-processamento, foi feito o one-hot-encoding das variáveis *Key*
e *Time Signature*. Também foi criado a variável *Popularity Class*,
definida como:

* 0–1: Unpopular
* 1–23: Low popularity
* 23–40: Moderate popularity
* > 40: High popularit

### Treinamento dos modelos
Foi feito o seguinte para treinar os modelos:
- 70% dos dados para treinamento e 30% para teste
- foi usado o *Standard Scaler* para escolar os dados (devido aos
  modelos lineares)
- 20 folds para validação cruzada
- Grid Search CV foi usado para tunar os hiperparâmetros

## Principais resultados
A floresta aleatória obteve os melhores resultados, tanto como um
modelo de regressão quanto um modelo de classificação. Os modelos de
classificação tiveram uma performance moderada, com escores na casa dos
0.6 (boa parte das métricas). Enquanto que os classificadores tiveram
escores muitos bons para músicas da classe 0 (não popular), com escores
na casa dos 0.8. Porém, para as demais classes de música, os
classificadores tiveram escores na casa dos 0.6.

> Observação:
> Suspeito que isso tenha ocorrido pois a classe das músicas não
> populares aparece com maior frequência.

O feature importance mostrou que o release_date (importância de 0,35)
tem grande importância no modelo, enquanto que as demais variáveis
(entre das 10 mais importantes) tem importância inferior a 0,05.

Também mostra que energy, valence, e instrumentalness são features
importantes para a predição de popularidade.
