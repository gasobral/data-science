# Anotações do artigo 1
**Título:** [Exploring Spotify's Music Popularity Dynamics and Forecasting with Machine Learning](https://www.researchgate.net/publication/385516062_Exploring_Spotify's_Music_Popularity_Dynamics_and_Forecasting_with_Machine_Learning)

**Autor:** Miaomiao Shu

**Ano:** 2024

## Objetivo
Analisar quais propriedades de uma música podem ser usadas para prever
a tendência/popularidade dela usando dados do *Spotify*.

## Metodologia
### Modelos utilizados
Nesse artigo são criados modelos usando *árvores de decisão* e *floresta
aleatória*

### Data sets
O autor menciona que o data set foi obtido do Kaggle, porém não coloca
o link no artigo. Encontrei os data sets abaixo que parecem ser os
dados utilizados no trabalho.

https://www.kaggle.com/datasets/nelgiriyewithana/top-spotify-songs-2023
https://www.kaggle.com/datasets/nelgiriyewithana/most-streamed-spotify-songs-2024

Não indica quais features foram usadas para gerar os dados da Tabela 2
(métricas de erro dos modelos), porém exibe um gráfico das features
importances (Figura 4). A partir desse gráfico, as 10 features de maior
importância são (listas em ordem descrente de importância):

![Figura 4: Feature Importances](figura_4_feature_importance.png)

1. in_spotify_playlists
2. in_deezer_playlists
3. in_apple_playlsits
4. released_year
5. in_spotify_charts
6. energy_%
7. valence_%
8. bpm
9. in_shazam_charts
10. in_appe_charts

Em particular, as duas primeiras têm um *feature importance* maior
igual a 0,3. Enquanto que as demais features têm uma importância menor
do que 0,1.

## Principais resultados
- músicas que adquirem alta popularidade numa plataforma tende a ser
  popular nas demais plataformas

- músicas com alta energia e valência positiva tendem a ter alta
  popularidade
