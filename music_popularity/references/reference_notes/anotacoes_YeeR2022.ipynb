# Anotações do artigo 5
**Título:** [Predicting Music Popularity Using Spotify and YouTube Features](https://www.academia.edu/108800370/Predicting_Music_Popularity_Using_Spotify_and_YouTube_Features)

**Autor:** Yap Kah Yee e Mafas Raheem

**Ano:** 2022

## Objetivo
Usar dados to YouTube e Spotify para prever a popularidade de uma
música. A ideia é analisar como os dados das redes sociais juntos com
as propriedades acústicas podem prever um hit.

## Metodologia
### Modelos utilizados
Random Forest, Logistic Regression, Support Vector Machine e Neural
Networks

### Data sets
Os dados foram obtidos das API's do Spotify e do Youtube de 14 de maio
a de 20 Agosto de 2021.

### Pipeline dos dados
![Pipe Line dos Dados](pipe_line_dados2.png)

### Treinamento dos modelos
Em relação a modelagem, as seguintes variáveis foram criadas:

- **Variáveis Independentes**: Length, Max, Sum, Mean e Debut
- **Variável Dependente**: popularidade = (baixa, média, alta)

Foram realizados dois experimentos para avaliar a performance dos
modelos:

1. Primeiro Experimento: usando apenas as features acústicas
2. Segundo Experimento: inclusão das features de redes sociais (número
   de votos, dislikes e o número de comentários)

Em ambos os experimentos os dados foram divididos em 80% treino e 20%
teste, 10 folds de validação cruzada e foi usado o Grid Search CV para
tunar os hiperparâmetros do modelo.

## Principais resultados
Sobre os experimentos, todos os modelos apresentaram melhoras
significativas ao adicionar as features vindas das redes sociais. A
floresta aleatória apresentou o melhor resultado geral.
