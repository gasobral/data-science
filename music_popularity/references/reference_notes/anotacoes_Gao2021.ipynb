# Anotações do artigo 3
**Título:** [Catching the Earworm: Understanding Streaming Music Popularity Using Machine Learning Models](https://www.e3s-conferences.org/articles/e3sconf/pdf/2021/29/e3sconf_eem2021_03024.pdf)

**Autor:** Andrea Gao

**Ano:** 2021

## Objetivo
Investigar a popularidade de músicas em plataformas de streaming,
utilizando informações sobre artistas e propriedades de músicas.

## Metodologia
O autor menciona que a popularidade de uma música em plataformas de
streaming tem uma mecânica diferente das vendas de música em CDs, DVDs
e outros. Inclusive, argumenta falando que a teoria da cauda longa
(*long tail theory*) foi rejeitada para o consumo de música em
plataformas de streaming.

O foco do trabalho é analisar aspectos acústicos da música para
predizer a popularidade de uma música.

> Observação: 
> No artigo é dito que a *composição* é a parte controlável do consumo
> da música.
>
> Ao analisar alguns artigos (referências 15, 16 e 17) sobre a
> popularidade de músicas, o autor menciona que um modelo linear pode
> não capturar bem a relação entre as features e a variável preditora
> (popularidade da música).

### Modelos utilizados
Multiple Linear Regression, Logistic Regression, Decision Tree, Random
Forest, and Boosting tree and Neural Networks.

### Data sets
No artigo é utilizado uma base de dados obtida do Spotify restrita aos
anos de 2018 a 2019, veja o link abaixo. Também é apresentado um
dicionário dos dados.

https://www.kaggle.com/datasets/yamaerenay/spotify-dataset-19212020-600k-tracks

Assim como na tese do Rusconi, a coluna popularidade é obtida por um
algoritmo interno do Spotify. Como não é fornencido muitos detalhes,
então os autores (desse artigo e Rusconi) suspeitam que o valor da
popularidade é computador a partir do volume de plays e do quão recente
uma música é.

### Limpeza dos dados
Foram removidas as músicas com popularidade = 0.

### Análise Exploratória
A partir do histograma da popularidade, essa métrica se concentra
bastante entre 0 (principalmente no 0) a 40, diminunido a sua
concentração de acordo que o valor de popularidade aumenta. Ao remover
as músicas com popularidade = 0, o histograma mostra um padrão em que a
frequência aumenta de acordo com a popularidade, de 0 a 35. Esse padrão
é chamado de *chasm*. Para construir um modelo que faça uma boa
classificação, é utilizado um *threshold* baseado no *chasm*. A ideia é
analisar as músicas a partir de um ponto em que popularidade decresce
com a frequência, produzindo um padrão de uma cauda longa. Para essa
finalidade é usado popularidade = 41 como *threshold*.

Ao analisar a correlação linear entre as variáveis, verificou-se que a
popularidade tem um correlação linear razoavelmente forte com
*loudness*. *Energy* possui uma correlação linear moderada com
popularidade, enquanto que *danceability* tem uma correlação um pouco
mais forte.

>Observação:
> A *popularidade* parece ter uma influência moderada no consumo de
> música. Ao menos ela tem uma correlação linear moderada, o que não
> exclui o fato de existir uma relação não linear. Sendo assim, seria
> interessante observar o feature significance numa floresta aleatória.

Também verificaram que o fato de um artista ter músicas na lista das
músicas mais populares não influencia diretamente o que ele irá
produzir no futuro.

### Treinamento dos modelos
Utilizaram a 0.005 de *learning rate*, o *GridSearchCV* para tunar os
hiperparâmetros e PCA para redução de dimensionalidade.

## Principais resultados
De modo geral, os modelos tiveram uma boa performance obtedo 0.8 em
boa tarde das métricas (acurário, F1 escore, precisão, revocação e
outras).

Os modelos mais simples (Regressão Linear Múltipla, Árovre de Decisão e
Regressão Logística) foram ligeiramente inferores ao modelos mais
complexos (Floresta Aleatória, Redes Neurais, Boosting Tree + PCA).

As features com importância (obtidos do feature importance) aos modelos
foram:
- loudness
- key_6
- duration_ms
- tempo
- liveness

A partir dos gráficos SHARP, temos o seguinte resultado:
- músicas com maior loudness tendem a serem mais populares
- notas F♯, G♭ tendem a estar presentes em músicas mais populares
- a *danceability* também tem uma influência positiva em classificar
  uma música como popular
