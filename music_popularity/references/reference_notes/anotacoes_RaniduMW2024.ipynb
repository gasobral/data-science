# Anotações do artigo 6
**Título:** [Stream Count Predictive Analysis for Upcoming Songs on Spotify using Machine Learning: A Systematic Literature Review](https://www.researchgate.net/publication/381581996_Stream_Count_Predictive_Analysis_for_Upcoming_Songs_on_Spotify_using_Machine_Learning_A_Systematic_Literature_Review)

**Autor:** M.G.Yoshitha Ranidu, Thilini V. Mahanama e Sankini Wijenarayana

**Ano:** 2024

## Objetivo
Revisão sistemática da literatura sobre a predição de músicas no
Spotify usando machine learning

## Principais resultados
Fatores que influenciam o trafego nas plataformas:

- popularidade
- layout da playlist
- marketing
- gênero
- conteúdo de redes sociais
- cultura
- dados do spotify

Resumo dos tipos de predições abordados e algoritmos de aprendizado de
máquina utilizados.

![Resumo dos tipo de predições](resumo_tipos_predicao.png)

Observações relevantes:
- usar dados regionais e levar em conta fatores culturais para fazer
  previsões com maior acurácia

- utilizar dados do Spotify para fazer previsões é essencial
