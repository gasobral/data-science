# Anotações do artigo 9
**Título:** [Predicting Afrobeats Hit Songs Using Spotify Data](https://arxiv.org/pdf/2007.03137)

**Autor:** Kai Middlebrook e Kian Sheik

**Ano:** 2019

## Objetivo
Previsão de qual música irá ser tornar popular (*Hit Song Science
Problem*) usando dados do Spotify.

## Metodologia

### Data sets
Foi construído um data set com aproximadamente 1,8 milhões de músicas
usando a API do Spotify e lista Hot 100 da Billboard, considerando o
período de 1985 a 2018.

### Modelos utilizados
- Logistic Regression
- Neural Network
- Random Forest
- Support Vector Machine

### Treinamento dos modelos
Foram utilizados features acústicas, tempo de duração e features do
artistas (obtidos dos dados da Billboard, mas não explicitamente
especificados no artigo).
